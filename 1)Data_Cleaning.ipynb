```python
import pandas as pd
import numpy as np

orders = pd.read_csv("1)Orders.csv")
customers = pd.read_csv("2) Customers.csv")
restaurants = pd.read_csv("3)Restuarant.csv")
delivery = pd.read_csv("4)Delivery.csv")
ratings = pd.read_csv("5)Ratings.csv")

print("All five datasets loaded successfully!")
```

```python
orders_clean = orders.copy()
customers_clean = customers.copy()
restaurants_clean = restaurants.copy()
delivery_clean = delivery.copy()
ratings_clean = ratings.copy()

print("All clean copies created successfully!")
```

```python
orders_clean["Order_Date_Time"] = pd.to_datetime(
    orders_clean["Order_Date_Time"],
    dayfirst=True,
    errors="coerce"
)

customers_clean["SignUp_Date"] = pd.to_datetime(
    customers_clean["SignUp_Date"],
    dayfirst=True,
    errors="coerce"
)

print("Date formatting completed!")
print(orders_clean.dtypes)
print(customers_clean.dtypes)
```

```python
orders_clean = orders.copy()
customers_clean = customers.copy()
restaurants_clean = restaurants.copy()
delivery_clean = delivery.copy()
ratings_clean = ratings.copy()
```

```python
orders_clean["Order_Date_Time"] = pd.to_datetime(
    orders_clean["Order_Date_Time"],
    dayfirst=True,
    errors="coerce"
)

customers_clean["SignUp_Date"] = pd.to_datetime(
    customers_clean["SignUp_Date"],
    dayfirst=True,
    errors="coerce"
)

print("Date formatting completed!")
print(orders_clean.dtypes)
print(customers_clean.dtypes)
```

```python
customers_clean["Customer_Name"] = (
    customers_clean["Customer_Name"].str.title()
)

customers_clean["City"] = (
    customers_clean["City"].str.title()
)

print("Customers standardized successfully!")

display(customers_clean.head())
```

```python
restaurants_clean["Restaurant_Name"] = (
    restaurants_clean["Restaurant_Name"].str.strip()
)

restaurants_clean["Cuisine"] = (
    restaurants_clean["Cuisine"].str.title()
)

restaurants_clean["City"] = (
    restaurants_clean["City"].str.title()
)

print("Restaurants standardized successfully!")

display(restaurants_clean.head())
```

```python
print("Restaurant Column Names:")
print(restaurants_clean.columns.tolist())
```

```python
restaurants_clean.columns = restaurants_clean.columns.str.strip()

print(restaurants_clean.columns.tolist())
```

```python
restaurants_clean["Restaurant_Name"] = restaurants_clean["Restaurant_Name"].str.strip().str.title()

restaurants_clean["Cuisine"] = restaurants_clean["Cuisine"].str.strip().str.title()

restaurants_clean["City"] = restaurants_clean["City"].str.strip().str.title()

print("Restaurant data standardized successfully!")
display(restaurants_clean.head())
```

```python
delivery_clean.columns = delivery_clean.columns.str.strip()

for col in delivery_clean.select_dtypes(include="object").columns:
    delivery_clean[col] = delivery_clean[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

delivery_clean["Delay_Status"] = (
    delivery_clean["Delay_Status"].str.title()
)

print("Delivery standardized successfully!")

display(delivery_clean.head())
```

```python
ratings_clean.columns = ratings_clean.columns.str.strip()

for col in ratings_clean.select_dtypes(include="object").columns:
    ratings_clean[col] = ratings_clean[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

ratings_clean["Feedback"] = (
    ratings_clean["Feedback"].str.title()
)

print("Ratings standardized successfully!")

display(ratings_clean.head())
```
